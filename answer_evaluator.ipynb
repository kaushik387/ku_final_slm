```python
# Load embedding model
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Load model
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# Evaluation function
def evaluate_answer(reference, student):

    ref_embedding = model.encode([reference])

    stu_embedding = model.encode([student])

    similarity = cosine_similarity(
        ref_embedding,
        stu_embedding
    )[0][0]

    score = round(similarity * 10, 2)

    return score
```

```python
# Test the function
reference = """
Clustering is an unsupervised learning technique that divides data into logical groups based on the characteristics of the data, where data within a cluster shares similarities. 
"""

student = """
Clustering is basically when you take a bunch of data and group the items that are similar together without any pre-made labels. It is like sorting a pile of random coins into separate stacks based on their size and color.
"""

score = evaluate_answer(
    reference,
    student
)

print("Score:", score)
```
