```python
from transformers import pipeline
from langchain_community.llms import HuggingFacePipeline
```

```python
pipe = pipeline(
    "text2text-generation",
    model="google/flan-t5-large",
    max_new_tokens=256,
    temperature=0.5
)

llm = HuggingFacePipeline(
    pipeline=pipe
)
```

```python
def generate_feedback(question,
                      reference_answer,
                      student_answer,
                      score):

    prompt = f"""
You are an AI tutor evaluating a student's answer.

Question:
{question}

Reference Answer:
{reference_answer}

Student Answer:
{student_answer}

Student Score:
{score}/10

Write feedback in the following format:

Correctness:
Write whether the answer is correct or partially correct.

Missing Concepts:
Mention important missing points.

Improvement Suggestions:
Suggest how the student can improve.

Final Feedback:
Give short encouraging educational feedback.

Do not repeat the reference answer.
"""

    result = llm.invoke(prompt)

    return result
```

```python
# Test the feedback generation
question = """
What is clustering in machine learning and why is it used?
"""

reference_answer = """
Clustering is an unsupervised machine learning technique
used to group similar data points into clusters based on
their patterns or similarities. It helps in discovering
hidden structures in data without using labeled outputs.
Common clustering algorithms include K-Means, DBSCAN,
and Hierarchical Clustering.
"""

student_answer = """
Clustering is a machine learning method that groups
similar data together. It is mainly used in unsupervised
learning to identify patterns and organize data into
different groups.
"""

score = 8.6

feedback = generate_feedback(
    question,
    reference_answer,
    student_answer,
    score
)

print("\n===== FEEDBACK =====\n")

for line in feedback.split("\n"):
    print(line)
```
